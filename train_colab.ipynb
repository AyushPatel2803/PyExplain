### 6. Push the trained adapter to your HuggingFace

Change `HF_USERNAME` below if your HuggingFace username is different.

```python
from huggingface_hub import HfApi, create_repo

HF_USERNAME = "AyushPatel28"
repo_id = f"{HF_USERNAME}/PyExplain-qwen-coder-7b"

create_repo(repo_id, exist_ok=True)
HfApi().upload_folder(folder_path="output", repo_id=repo_id)
print("Done! Your model is at: https://huggingface.co/" + repo_id)
```

## Done!

Your fine-tuned **7B** adapter is now on HuggingFace. To use it locally with
PyExplain, set `MODEL_NAME = "Qwen/Qwen2.5-Coder-7B-Instruct"` in `config.py`
(already the default) and point `OUTPUT_DIR` at the downloaded adapter, or load
it directly from the HuggingFace repo.
